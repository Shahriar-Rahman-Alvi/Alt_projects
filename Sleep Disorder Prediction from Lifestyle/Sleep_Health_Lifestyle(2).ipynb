## **3.  Machine Learning Model for Sleep Disorder Prediction**


*   **Setting up the environment**

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, f_classif, RFE
```


*   ****


*   **Separate features and target**





```python
X = df.iloc[:, :-1] # Make a list of Features
y = df.iloc[:, -1]  # Make a list of Target Variables
```

```python
X.shape
```

```python
y.shape
```

```python
print(y.value_counts(normalize=True)*100)
```

                                  Here, "No Sleep Disorder" dominates. So, the dataFrame is imbalance.

```python
# encode all string features
numeric_features = X.select_dtypes(include=["int64", "float64"]).columns
categorical_features = X.select_dtypes(include=["object"]).columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

*   **Split the data into Train-&-Test**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state = 99, test_size = 0.2
) # in the "random_state", we can put any number here (sets a fixed seed).
```


###   **Model Random Forest**


```python
#Number of trees (n_estimators): the number of decision trees to combine
#Maximum depth (max_depth): the maximum depth of each decision tree
#random_state sets a fixed seed to ensure the same output every time the code is run

randomForest = Pipeline(steps=[
                        ("preprocessing", preprocessor),
                        ("smote", SMOTE(random_state=99)),
                        ("classifier", RandomForestClassifier(
                                                    n_estimators=50,
                                                    random_state=67
                       ))
])
randomForest.fit(X_train, y_train)
```


*   **Make predictions**


```python
y_pred_rf = randomForest.predict(X_test)
y_pred_rf
```


*   **Confusion Matrix**


```python
cm_rf = confusion_matrix(y_test, y_pred_rf)
cm_rf
```

```python
sns.heatmap(cm_rf, annot=True, cmap='Blues', fmt='d')
plt.title("Confusion Matrix - Random Forest")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```


*   **Accuracy**


```python
# Model Random Forest (RF)
rf_accuracy = accuracy_score(y_test, y_pred_rf)
print("Random Forest Accuracy:", rf_accuracy*100,"%")
```


*   **Cross Validation**


```python
# Doing cross validation 10 times
cross_val_score(randomForest, X_train, y_train, cv = 10)
```



*   **Random Forest Classification report**



```python
print(classification_report(y_test, y_pred_rf))
```

### **Model Logistic Regression**

```python
lr=Pipeline(steps=[
                ("preprocessing", preprocessor),
                ("smote", SMOTE(random_state=99)),
                ("classifier", LogisticRegression(
                    C=1,
                    random_state=99
                ))
])
lr.fit(X_train, y_train)
```

*  **Make predictions**

```python
y_pred_lr = lr.predict(X_test)
y_pred_lr
```

*   **Confusion Matrix**

```python
cm_lr = confusion_matrix(y_test, y_pred_lr)
cm_lr
```

```python
sns.heatmap(cm_lr, annot=True, cmap='Blues', fmt='d')
plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

*   **Accuracy**

```python
lr_accuracy = accuracy_score(y_test, y_pred_lr)
print("Logistic Regression Accuracy:", lr_accuracy*100,"%")
```

*   **Cross Validation**

```python
cross_val_score(lr, X_train, y_train, cv = 10)
```

*   **Logistic Regression Classification report**

```python
print(classification_report(y_test, y_pred_lr))
```

### **Model Decision Tree**

```python
dec_tree=Pipeline(steps=[
                        ("preprocessing", preprocessor),
                        ("smote", SMOTE(random_state=99)),
                        ("classifier", DecisionTreeClassifier())
])
dec_tree.fit(X_train, y_train)
```

*   **Make predictions**

```python
y_pred_dec_tree = dec_tree.predict(X_test)
y_pred_dec_tree
```

*   **Confusion Matrix**

```python
cm_dec_tree = confusion_matrix(y_test, y_pred_dec_tree)
cm_dec_tree
```

```python
sns.heatmap(cm_dec_tree, annot=True, cmap='Blues', fmt='d')
plt.title("Confusion Matrix - Decision Tree")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

*   **Accuracy**


```python
dec_tree_accuracy = accuracy_score(y_test, y_pred_dec_tree)
print("Decision Tree Accuracy:", dec_tree_accuracy*100,"%")
```

*   **Cross Validation**

```python
cross_val_score(dec_tree, X_train, y_train, cv = 10)
```

*   **Decision Tree Classification report**

```python
print(classification_report(y_test, y_pred_dec_tree))
```
